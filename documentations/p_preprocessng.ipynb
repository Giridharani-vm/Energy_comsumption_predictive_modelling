```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv(r"C:\prj_power_consumption\Cleaned_data_eda.csv")
df['Date'] = pd.to_datetime(df['Date'])
df['Time'] = pd.to_datetime(df['Time']).dt.time
df.dtypes
```

```python
df.size
df.shape
```

```python
df.isnull().sum()

```

```python
df['Datetime'] = pd.to_datetime(df['Date'].astype(str) + ' ' + df['Time'].astype(str))
df['Date_only'] = df['Datetime'].dt.date

# Calculate daily average Global Active Power
daily_avg = df.groupby('Date_only')['Global_active_power'].mean().rename('daily_avg_power')

# Calculate daily peak hour (hour with max power)
peak_hour = df.groupby(['Date_only', 'Hour'])['Global_active_power'].mean().reset_index()
peak_hour = peak_hour.loc[peak_hour.groupby('Date_only')['Global_active_power'].idxmax()]
peak_hour = peak_hour.set_index('Date_only')['Hour'].rename('daily_peak_hour')

# Merge back to original dataframe
df = df.merge(daily_avg, left_on='Date_only', right_index=True)
df = df.merge(peak_hour, left_on='Date_only', right_index=True)

# Display one day sample
df[df['Date_only'] == df['Date_only'].unique()[0]]
```

```python
df[25:100]
```

```python
# Log transform the skewed features
df['Global_active_power'] = np.log1p(df['Global_active_power'])
df['Global_reactive_power'] = np.log1p(df['Global_reactive_power'])
df['Global_intensity'] = np.log1p(df['Global_intensity'])
df['Sub_metering_1'] = np.log1p(df['Sub_metering_1'])
df['Sub_metering_2'] = np.log1p(df['Sub_metering_2'])

# Apply StandardScaler
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df[['Global_active_power', 'Global_reactive_power','Voltage', 'Global_intensity','Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3']])

# Convert back to DataFrame if needed
scaled_df = pd.DataFrame(scaled_features, columns=['Global_active_power', 'Global_reactive_power',
                                                    'Voltage', 'Global_intensity',
                                                    'Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3'])
```
