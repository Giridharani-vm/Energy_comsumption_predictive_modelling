```python
import matplotlib.pyplot as plt

# R² Scores
models = [
    "Linear Regression (Ridge Optimized)",
    "Random Forest regressor (Tuned)",
    "Gradient Boosting regressor  (GridSearchCV)",
    "MLP neural network"
]
scores = [0.7050, 0.6727, 0.6877, 0.6633]

# Plotting
plt.figure(figsize=(10, 6))
bars = plt.bar(models, scores, color=['skyblue', 'lightgreen', 'salmon', 'plum'])
plt.ylim(0.6, 0.75)
plt.title("R² Score Comparison Across Regression Models", fontsize=14)
plt.ylabel("R² Score")
plt.xticks(rotation=15, ha='right')

# Adding text labels on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, height, f'{height:.4f}', ha='center', va='bottom')

plt.tight_layout()
plt.show()
```
